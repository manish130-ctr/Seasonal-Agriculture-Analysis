```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv(r"C:\Users\maity\Downloads\Telegram Desktop\seasonal_agriculture_performance_dataset (2).csv")
df.head()
```

```python

cols_to_fill = ['Rainfall_mm', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha']
for col in cols_to_fill:
    df[col] = df[col].fillna(df[col].median())

print("Missing values after cleaning:\n", df.isnull().sum())
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Season', y='Water_Used_m3', palette='Set2')
plt.title('Seasonal Variations in Water Usage')
plt.ylabel('Water Used (m³)')
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Season', y='Profit_INR', errorbar=None, palette='Set2')

plt.title('Average Profit by Season')
plt.xlabel('Season')
plt.ylabel('Average Profit (INR)')
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.barplot(data=df, x='Season', y='Water_Used_m3', errorbar=None, palette='mako')

plt.yticks(range(0, 7500, 500))

plt.title('Average Water Usage by Season')
plt.xlabel('Season')
plt.ylabel('Water Used (m³)')
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Season', y='Disease_Pest_Risk_pct', errorbar=None, palette='Reds')
plt.title('Average Disease and Pest Risk by Season')
plt.xlabel('Season')
plt.ylabel('Pest/Disease Risk (%)')
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Season', y='Yield_Tonnes_Ha', errorbar=None, palette='YlGnBu')
plt.title('Average Crop Yield by Season')
plt.xlabel('Season')
plt.ylabel('Yield (Tonnes per Hectare)')
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x='Total_Cost_INR', y='Revenue_INR', hue='Season', alpha=0.7, palette='deep')
plt.title('Total Cost vs Revenue by Season', fontsize=14, fontweight='bold')
plt.xlabel('Total Cost (INR)', fontsize=12)
plt.ylabel('Revenue (INR)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```
